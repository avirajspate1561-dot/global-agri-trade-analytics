# Global Agricultural Trade & Risk Analytics Dashboard

## Project Objective

This project analyzes agricultural trade flows, commodity prices,
trade balances, and price volatility across major countries and
agricultural commodities from 2015 to 2025.

The project combines:

- Python data generation
- Pandas data preprocessing
- Statistical analysis
- Tableau visualization
- Geographic analysis
- Forecasting
- Executive reporting
- GitHub portfolio publishing

## Commodities

- Wheat
- Maize
- Soybeans
- Rice

## Countries

- United States
- Brazil
- Argentina
- Ukraine
- China
- India
- Egypt
- Germany

```python
# Cell 2: Generate Global Agricultural Trade Dataset

import os
import numpy as np
import pandas as pd

# --------------------------------------------------
# 1. Reproducibility
# --------------------------------------------------

np.random.seed(42)


# --------------------------------------------------
# 2. Parameters
# --------------------------------------------------

years = list(range(2015, 2026))

countries = [
    "United States",
    "Brazil",
    "Argentina",
    "Ukraine",
    "China",
    "India",
    "Egypt",
    "Germany"
]

commodities = [
    "Wheat",
    "Maize",
    "Soybeans",
    "Rice"
]


# --------------------------------------------------
# 3. Generate Records
# --------------------------------------------------

records = []

for year in years:

    for country in countries:

        for commodity in commodities:

            # Exporter countries
            is_exporter = country in [
                "United States",
                "Brazil",
                "Argentina",
                "Ukraine"
            ]

            # Export volume
            if is_exporter:
                export_vol = np.random.randint(
                    5000,
                    25000
                )
            else:
                export_vol = np.random.randint(
                    100,
                    3000
                )

            # Import volume
            if is_exporter:
                import_vol = np.random.randint(
                    100,
                    3000
                )
            else:
                import_vol = np.random.randint(
                    5000,
                    22000
                )

            # Commodity base prices
            base_price = {
                "Wheat": 220,
                "Maize": 180,
                "Soybeans": 400,
                "Rice": 350
            }[commodity]

            # Price trend
            price_usd = round(
                base_price
                * (1 + (year - 2015) * 0.05)
                + np.random.uniform(-20, 30),
                2
            )

            # Volatility
            price_volatility = round(
                np.random.uniform(5.0, 25.0),
                2
            )

            # Net trade balance
            net_trade_balance = (
                export_vol - import_vol
            )

            records.append({

                "Year": year,

                "Country": country,

                "Commodity": commodity,

                "Export_Volume_MT": export_vol,

                "Import_Volume_MT": import_vol,

                "Price_USD_per_MT": price_usd,

                "Price_Volatility_Index": price_volatility,

                "Net_Trade_Balance_MT": net_trade_balance
            })


# --------------------------------------------------
# 4. Create DataFrame
# --------------------------------------------------

df = pd.DataFrame(records)


# --------------------------------------------------
# 5. Calculate Trade Values
# --------------------------------------------------

df["Export_Value_USD"] = (
    df["Export_Volume_MT"]
    * df["Price_USD_per_MT"]
)

df["Import_Value_USD"] = (
    df["Import_Volume_MT"]
    * df["Price_USD_per_MT"]
)


# --------------------------------------------------
# 6. Create Output Directory
# --------------------------------------------------

os.makedirs(
    "../data",
    exist_ok=True
)


# --------------------------------------------------
# 7. Save Dataset
# --------------------------------------------------

output_path = (
    "../data/global_agricultural_trade.csv"
)

df.to_csv(
    output_path,
    index=False
)


# --------------------------------------------------
# 8. Display Information
# --------------------------------------------------

print(
    "Dataset successfully generated:"
)

print(output_path)

print(
    f"\nRows: {len(df):,}"
)

print(
    f"Columns: {len(df.columns)}"
)

display(df.head())
```

```python
# Cell 3: Data Quality Checks

print("Shape:")
print(df.shape)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nData Types:")
print(df.dtypes)

print("\nYears:")
print(
    df["Year"].min(),
    "to",
    df["Year"].max()
)

print("\nCountries:")
print(
    df["Country"].unique()
)

print("\nCommodities:")
print(
    df["Commodity"].unique()
)
```

```python
# Cell 4: Statistical Summary

summary = df[
    [
        "Export_Volume_MT",
        "Import_Volume_MT",
        "Price_USD_per_MT",
        "Price_Volatility_Index",
        "Net_Trade_Balance_MT",
        "Export_Value_USD",
        "Import_Value_USD"
    ]
].describe()

display(summary)
```

```python
# Cell 5: Global Trade Balance

global_export_value = df[
    "Export_Value_USD"
].sum()

global_import_value = df[
    "Import_Value_USD"
].sum()

global_trade_balance = (
    global_export_value
    - global_import_value
)

global_trade_balance_ratio = (
    global_trade_balance
    / global_import_value
)

print(
    f"Global Export Value: "
    f"${global_export_value:,.2f}"
)

print(
    f"Global Import Value: "
    f"${global_import_value:,.2f}"
)

print(
    f"Global Trade Balance: "
    f"${global_trade_balance:,.2f}"
)

print(
    f"Trade Balance Ratio: "
    f"{global_trade_balance_ratio:.2%}"
)
```

```python
# Cell 6: Country Trade Risk Analysis

country_analysis = (
    df.groupby("Country")
    .agg(
        Total_Exports_MT=(
            "Export_Volume_MT",
            "sum"
        ),

        Total_Imports_MT=(
            "Import_Volume_MT",
            "sum"
        ),

        Total_Export_Value_USD=(
            "Export_Value_USD",
            "sum"
        ),

        Total_Import_Value_USD=(
            "Import_Value_USD",
            "sum"
        ),

        Average_Volatility=(
            "Price_Volatility_Index",
            "mean"
        )
    )
    .reset_index()
)

country_analysis[
    "Net_Trade_Balance_MT"
] = (
    country_analysis["Total_Exports_MT"]
    - country_analysis["Total_Imports_MT"]
)

country_analysis[
    "Import_Dependency_Ratio"
] = (
    country_analysis["Total_Imports_MT"]
    /
    country_analysis["Total_Exports_MT"]
)

display(
    country_analysis.sort_values(
        "Net_Trade_Balance_MT"
    )
)
```

```python
# Cell 7: Volatility Change

yearly_volatility = (
    df.groupby("Year")[
        "Price_Volatility_Index"
    ]
    .mean()
    .reset_index()
)

volatility_2015 = yearly_volatility.loc[
    yearly_volatility["Year"] == 2015,
    "Price_Volatility_Index"
].iloc[0]

volatility_2025 = yearly_volatility.loc[
    yearly_volatility["Year"] == 2025,
    "Price_Volatility_Index"
].iloc[0]

volatility_change = (
    (volatility_2025 - volatility_2015)
    / volatility_2015
) * 100

print(
    f"2015 Average Volatility: "
    f"{volatility_2015:.2f}"
)

print(
    f"2025 Average Volatility: "
    f"{volatility_2025:.2f}"
)

print(
    f"Change: "
    f"{volatility_change:.2f}%"
)
```

```python
# Cell 8: Correlation Analysis

correlation = df[
    [
        "Export_Volume_MT",
        "Import_Volume_MT",
        "Price_USD_per_MT",
        "Price_Volatility_Index"
    ]
].corr()

display(correlation)
```

```python
# Cell 9: Save Country Analysis

country_analysis.to_csv(
    "../data/country_trade_risk_analysis.csv",
    index=False
)

yearly_volatility.to_csv(
    "../data/yearly_volatility.csv",
    index=False
)

print("Analysis files saved.")
```
